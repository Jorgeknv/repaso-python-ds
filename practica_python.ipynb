```python
# Datos del producto
nombre_producto = "Mouse"
precio = 85.50
stock = 12
tiene_descuento = True

descuento = "sí" if tiene_descuento else "no"

print(
    f"El producto {nombre_producto} cuesta ${precio:.2f}, "
    f"tiene {stock} unidades en stock. Descuento disponible: {descuento}."
)
```

```python
# Clasificación de Precios
precios = [49.90, 99.90, 100, 100.01, 120.50]

for precio in precios:
    if precio > 100:
        print (f"$ {precio:.2f}: Caro")
    else:
        print (f"$ {precio:.2f}: Económico")
```

```python
#Registro y consulta del stock
almacen = {"Teclado": 15, "Mouse": 12, "Auriculares": 30}

almacen["Webcam"] = 8

print(f"Stock de Webcam: {almacen['Webcam']} unidades.")
```

```python
# Resumen de una lista de números
def resumen_estadistico(numeros):
    total = sum(numeros)
    cantidad = len(numeros)
    promedio = total/cantidad if cantidad else 0.0

    return {"suma": total, "promedio": promedio, "cantidad": cantidad}

print(resumen_estadistico([10, 20, 30]))
```
